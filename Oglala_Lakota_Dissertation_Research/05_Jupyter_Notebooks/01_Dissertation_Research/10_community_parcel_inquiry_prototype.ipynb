
# SA 18.9 — Community Parcel Data Cube Inquiry Prototype

## Environmental Data Cube for Oglala Lakota
### Community-Governed Environmental Data Services

**Concept:** A Data Sovereignty Steward assists community members
in requesting, developing, and interpreting environmental data
cubes for individual land parcels.

**Development status:** Conceptual prototype — not a deployed
website or operational data service.

---

## 1. Purpose

The Community Parcel Data Cube Inquiry Prototype explores how
the Environmental Data Cube for Oglala Lakota could support
community members seeking environmental information about
their land.

A community member may approach a Data Sovereignty Steward
with a question such as:

> "I have a grazing parcel on Pine Ridge. I know the township,
> range, and section. Can you help me understand how the land
> has changed over time?"

The proposed service would provide a community-governed
process for requesting a parcel-specific environmental
assessment while respecting Tribal Data Sovereignty,
individual privacy, and applicable land governance protocols.

The inquiry website would serve as an initial request and
stewardship interface rather than a public parcel database.

---

## 2. Community Inquiry Form

### A. Community Member Information

The proposed form would request:

- Name and preferred contact method
- Relationship to the parcel
- Landholder, landowner, leaseholder, permit holder, or
  authorized representative status
- Purpose of the environmental assessment

### B. Parcel Location

Community members could provide a legal land description
using the Public Land Survey System (PLSS):

- Township number and direction
- Range number and direction
- Section number
- Quarter-section or other subdivision, if known
- Principal meridian, if known
- Additional parcel or tract information, when appropriate

Township and range alone do not identify an individual parcel.
The Steward would verify the legal description and parcel
boundary using appropriate land records and permissions.

A future version could also accept an authorized parcel
Shapefile, GeoJSON, or geographic bounding box.

### C. Research Questions

Community members could select their areas of interest:

1. Grazing and rangeland conditions
2. Vegetation change over time
3. Drought and precipitation history
4. Land restoration and conservation planning
5. Other environmental research questions

The form would also provide space for the member to describe
their concerns and desired outcomes.

---

## 3. Data Sovereignty Steward Workflow

### Step 1 — Receive the Inquiry

The Data Sovereignty Steward receives the community member's
request, clarifies its purpose, and explains the proposed
handling of parcel information.

### Step 2 — Verify the Parcel

The Steward confirms the legal land description, applicable
permissions, and authorized parcel boundary.

Parcel location information would not be published
automatically.

### Step 3 — Develop the Parcel Data Cube

Where suitable source data are available, the Steward would
prepare a parcel-specific environmental dataset.

Potential variables include:

- NDVI — vegetation greenness
- EVI — enhanced vegetation index
- Precipitation
- Temperature
- Additional rangeland indicators, where scientifically
  justified and supported by available data

The data cube would document:

- Geographic extent
- Temporal coverage
- Satellite and climate data sources
- Spatial and temporal resolution
- Processing methodology
- Quality screening
- Uncertainty and limitations

A 2000–2023 historical assessment could be considered where
data availability and comparability permit.

The existing annual community data cube cannot, by itself,
produce parcel-specific measurements for arbitrary parcels.
Spatially referenced imagery or other suitable spatial data
would be required.

### Step 4 — Interpret Environmental Conditions

The Steward and community member review environmental
patterns, including vegetation changes and climate conditions.

Vegetation declines would not automatically be interpreted
as overgrazing.

Possible explanations include drought, seasonal variability,
land management, soil conditions, and other ecological factors.

### Step 5 — Deliver the Assessment

Potential private deliverables include:

- Parcel boundary visualization
- Environmental time-series charts
- Vegetation and climate summaries
- Quality and uncertainty documentation
- Parcel-specific NetCDF data cube
- Plain-language environmental assessment

The community member would receive information in a format
appropriate to the request and authorized uses.

---

## 4. Tribal Data Sovereignty Framework

The prototype is guided by Tribal Data Sovereignty and
community-defined governance.

Before operational deployment, the service would require
approved procedures addressing:

- Verification of authority to request parcel information
- Informed consent and permitted uses
- Access restrictions for parcel boundaries and records
- Secure storage and data retention
- Data correction and deletion procedures
- Restrictions on external sharing
- Separate authorization for research reuse or publication
- Applicable Tribal governance and cultural protocols

The service would distinguish between environmental data
generated for an individual parcel and data authorized
for broader research or public distribution.

No parcel-specific information would be publicly released
without appropriate authorization.

---

## 5. Proposed Website Structure

### Page 1 — Welcome

**Request an Environmental Assessment for Your Land**

A community-focused introduction explaining the purpose
of the service and the role of the Data Sovereignty Steward.

### Page 2 — Community Member Inquiry

Collects contact information, relationship to the land,
and the purpose of the request.

### Page 3 — Locate Your Parcel

Accepts township, range, section, and subdivision
information, with optional geographic data where authorized.

### Page 4 — Select Environmental Questions

Allows the community member to identify vegetation,
grazing, drought, or restoration concerns.

### Page 5 — Review and Consent

Explains how the information would be handled and provides
a process for submitting the inquiry.

### Page 6 — Private Environmental Explorer

After verification and analysis, an authorized member could
review parcel-specific charts, environmental summaries,
and available data products.

## 7. Development Boundaries

This is a proposed research and service design.

The prototype does not currently:

- Operate as a live public website
- Verify land ownership or lease authority
- Convert legal descriptions into verified parcel boundaries
- Generate new parcel-specific satellite data cubes
  automatically
- Provide an approved operational consent or data
  governance system

These functions would require further development,
technical validation, and appropriate governance review.

```python

# SA 18.9 — Local, read-only demonstration interface
# This cell does not access your research datasets or write any files.

import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown
import html

style = {"description_width": "165px"}
layout = widgets.Layout(width="520px")

member = widgets.Text(
    description="Member (fictional):",
    placeholder="Example Person",
    style=style, layout=layout
)

contact = widgets.Text(
    description="Contact (fictional):",
    placeholder="Do not enter real contact details",
    style=style, layout=layout
)

relationship = widgets.Dropdown(
    description="Relationship:",
    options=[
        "Landholder / landowner",
        "Leaseholder / permit holder",
        "Authorized representative",
        "Other / not yet verified"
    ],
    style=style, layout=layout
)

township = widgets.Text(
    description="Township:",
    placeholder="Example: 38",
    style=style, layout=layout
)

township_dir = widgets.Dropdown(
    description="Township direction:",
    options=["N", "S"],
    style=style, layout=layout
)

range_num = widgets.Text(
    description="Range:",
    placeholder="Example: 45",
    style=style, layout=layout
)

range_dir = widgets.Dropdown(
    description="Range direction:",
    options=["W", "E"],
    style=style, layout=layout
)

section = widgets.Text(
    description="Section:",
    placeholder="1–36",
    style=style, layout=layout
)

subdivision = widgets.Text(
    description="Subdivision:",
    placeholder="Optional, e.g. NW 1/4",
    style=style, layout=layout
)

meridian = widgets.Text(
    description="Principal meridian:",
    placeholder="To be verified",
    style=style, layout=layout
)

interest = widgets.SelectMultiple(
    description="Questions:",
    options=[
        "Grazing and rangeland conditions",
        "Vegetation change",
        "Drought and precipitation",
        "Restoration planning"
    ],
    value=("Vegetation change",),
    style=style,
    layout=widgets.Layout(width="620px", height="120px")
)

notes = widgets.Textarea(
    description="Additional context:",
    placeholder="Synthetic demonstration text only",
    style=style,
    layout=widgets.Layout(width="620px", height="90px")
)

ack = widgets.Checkbox(
    value=False,
    description=(
        "I understand this is a synthetic, "
        "non-submitting demonstration."
    ),
    indent=False,
    layout=widgets.Layout(width="650px")
)

review_btn = widgets.Button(
    description="Review sample inquiry",
    button_style="info",
    icon="search"
)

preview_btn = widgets.Button(
    description="Show synthetic cube preview",
    button_style="success",
    icon="bar-chart"
)

output = widgets.Output()


# ---------------------------------------------------
# VALIDATE SAMPLE LAND DESCRIPTION
# ---------------------------------------------------

def valid_description():

    if (
        not township.value.strip()
        or not range_num.value.strip()
        or not section.value.strip()
    ):
        return "Enter a sample township, range, and section."

    if (
        not township.value.strip().isdigit()
        or not range_num.value.strip().isdigit()
    ):
        return "Township and range must be positive whole numbers."

    if int(township.value) < 1 or int(range_num.value) < 1:
        return "Township and range must be positive."

    if (
        not section.value.strip().isdigit()
        or not 1 <= int(section.value) <= 36
    ):
        return "Section must be a whole number from 1 to 36."

    if not ack.value:
        return "Check the demonstration acknowledgement first."

    return None


# ---------------------------------------------------
# STEWARD REVIEW
# ---------------------------------------------------

def on_review(_):

    with output:
        clear_output(wait=True)

        error = valid_description()

        if error:
            display(Markdown("**Review needed:** " + error))
            return

        description = (
            f"T{int(township.value)}{township_dir.value} "
            f"R{int(range_num.value)}{range_dir.value}, "
            f"Section {int(section.value)}"
        )

        if subdivision.value.strip():
            description += ", " + subdivision.value.strip()

        fields = [
            ("Member (fictional)", member.value or "Not supplied"),
            ("Contact (fictional)", contact.value or "Not supplied"),
            ("Relationship (unverified)", relationship.value),
            ("Legal description (unverified)", description),
            (
                "Principal meridian",
                meridian.value or "Not provided — verify"
            ),
            (
                "Research interests",
                ", ".join(interest.value) or "Not selected"
            ),
            ("Additional context", notes.value or "None")
        ]

        rows = "".join(
            "<tr><th style='text-align:left;padding:6px'>"
            + html.escape(k)
            + "</th><td style='padding:6px'>"
            + html.escape(v)
            + "</td></tr>"
            for k, v in fields
        )

        display(widgets.HTML(
            "<h3>Steward review — sample inquiry</h3>"
            "<table>" + rows + "</table>"
            "<p><strong>Status:</strong> Draft only; "
            "no request submitted or saved.</p>"
        ))

        display(Markdown(
            "**Before any real analysis:** Verify requester "
            "authority, legal description, principal meridian, "
            "actual parcel geometry, permissions, and "
            "approved data handling."
        ))


# ---------------------------------------------------
# SYNTHETIC DATA CUBE PREVIEW
# ---------------------------------------------------

def on_preview(_):

    with output:
        clear_output(wait=True)

        if not ack.value:
            display(Markdown(
                "Check the demonstration acknowledgement first."
            ))
            return

        display(Markdown(
            "### Illustrative private parcel data cube "
            "— **synthetic values**"
        ))

        import pandas as pd

        synthetic = pd.DataFrame({
            "year": [2019, 2020, 2021, 2022, 2023],
            "ndvi_demo": [0.52, 0.46, 0.34, 0.40, 0.50],
            "precip_mm_demo": [410, 325, 245, 290, 390],
            "tmax_k_demo": [297.0, 298.1, 300.0, 299.2, 297.6]
        }).set_index("year")

        display(synthetic)

        try:
            import matplotlib.pyplot as plt

            ax = synthetic["ndvi_demo"].plot(
                marker="o",
                figsize=(7, 3.2),
                title="SYNTHETIC NDVI — not a real parcel"
            )

            ax.set_ylabel("Illustrative NDVI")
            ax.set_xlabel("Year")

            plt.tight_layout()
            plt.show()

        except ImportError:
            display(Markdown(
                "Matplotlib is not installed; "
                "the synthetic table above is sufficient."
            ))

        display(Markdown(
            "**Not derived from your community cube or any "
            "land parcel.** A future authorized implementation "
            "could create a spatially derived NetCDF and a "
            "private report, with documented quality "
            "and uncertainty."
        ))


# ---------------------------------------------------
# CONNECT BUTTONS
# ---------------------------------------------------

review_btn.on_click(on_review)
preview_btn.on_click(on_preview)


# ---------------------------------------------------
# DISPLAY COMMUNITY INQUIRY INTERFACE
# ---------------------------------------------------

display(widgets.VBox([

    widgets.HTML(
        "<h2>Community Parcel Data Cube Inquiry</h2>"
        "<p><b>Environmental Data Cube for Oglala Lakota</b>"
        "<br>Local, fictional demonstration only</p>"
    ),

    widgets.HTML("<h3>1. Community member</h3>"),
    member,
    contact,
    relationship,

    widgets.HTML(
        "<h3>2. Legal land description</h3>"
        "<p>Township/range/section are not yet "
        "a verified parcel boundary.</p>"
    ),

    township,
    township_dir,
    range_num,
    range_dir,
    section,
    subdivision,
    meridian,

    widgets.HTML("<h3>3. Environmental inquiry</h3>"),
    interest,
    notes,

    widgets.HTML("<h3>4. Steward review</h3>"),
    ack,

    widgets.HBox([review_btn, preview_btn]),

    output

]))
```
