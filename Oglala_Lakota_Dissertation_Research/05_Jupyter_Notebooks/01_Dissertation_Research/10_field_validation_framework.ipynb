## 1. Research Framework

The purpose of this research framework is to connect satellite-derived vegetation-change observations with independent ecological field measurements and authorized grazing-management information. This integration is intended to support a scientifically rigorous and culturally appropriate assessment of agricultural sustainability.

The framework consists of five interconnected stages:

### Stage 1 — Environmental Data Cube

**Existing research evidence:**
- Landsat-derived NDVI and NDMI vegetation indices.
- Vegetation-change trajectories for 2019, 2022, and 2023.
- Original growing-season and July–August composite comparisons.
- Climate context, including temperature and precipitation.
- Spatial classification of vegetation decline and recovery.

**Purpose:** Identify spatial and temporal patterns of vegetation condition and change within the study parcel.

### Stage 2 — Provisional Field-Validation Sampling

**Sampling framework:**
- 30 provisional field-validation candidates.
- Five vegetation-response strata.
- Six candidates allocated to each original stratum.
- Minimum spatial separation and parcel-boundary clearance evaluated.
- Seasonal classification sensitivity documented.

**Purpose:** Establish spatially distributed candidate locations for independent ecological observations.

The sampling locations remain provisional and are not authorized field sites.

### Stage 3 — Ecological Field Indicators

| Domain | Proposed indicator | Measurement approach |
|---|---|---|
| Vegetation | Vegetation cover (%) | Line-point intercept |
| Vegetation | Aboveground forage biomass (kg dry matter/ha) | Quadrat sampling and dry-weight measurement |
| Vegetation | Plant species composition | Species-level field survey |
| Soil | Bare ground (%) | Line-point intercept |
| Soil | Soil surface condition | Standardized surface assessment |
| Soil | Visible erosion indicators | Rill, pedestal, and soil movement observations |
| Grazing | Forage utilization (%) | Approved utilization protocol |
| Management | Stocking and grazing history | Authorized management records |

**Purpose:** Provide independent ecological evidence to evaluate and contextualize remotely sensed vegetation patterns.

These measurements are proposed; field observations have not yet been collected or integrated.

### Stage 4 — Tribal Data Sovereignty and Research Governance

Research activities must incorporate appropriate Tribal governance, cultural protocols, and data stewardship requirements.

**Key considerations:**
- Obtain appropriate Tribal authorization before field activities.
- Establish approved protocols for collecting ecological and management information.
- Determine permissions for accessing and using grazing-management records.
- Establish requirements for data storage, access, analysis, publication, and sharing.
- Protect sensitive land-management and culturally significant information.
- Incorporate community-approved ecological reference conditions and assessment criteria.

**Purpose:** Ensure that research methods and data practices respect the rights, authority, and priorities of the Oglala Lakota Nation.

### Stage 5 — Integrated Agricultural Sustainability Assessment

The final assessment will integrate:

1. **Remote sensing evidence:** Vegetation-index changes, spatial patterns, and recovery trajectories.
2. **Climate context:** Precipitation and temperature conditions associated with the observation periods.
3. **Ecological field evidence:** Vegetation cover, forage biomass, species composition, and soil condition.
4. **Management context:** Authorized grazing records and land-management information.
5. **Research governance:** Tribal authorization, data stewardship, and approved assessment criteria.

**Purpose:** Develop an evidence-based understanding of agricultural sustainability that considers vegetation dynamics, ecological conditions, climate variability, and land-management context.

An overall agricultural sustainability rating cannot be assigned until sufficient independent ecological evidence, management information, and appropriate reference criteria are available.

## 4. Methodological Considerations

The remote-sensing analyses demonstrate consistent parcel-average vegetation-index decline from 2019 to 2022, followed by partial recovery in 2023. However, discrete vegetation-response classifications exhibit sensitivity to seasonal compositing choices.

These findings establish the importance of independent field observations and uncertainty evaluation.

NDVI and NDMI changes should not be interpreted as direct evidence of overgrazing, livestock impacts, or agricultural sustainability without supporting ecological and management information.

Furthermore, the provisional stratified candidate sample is not a probability sample and should not be treated as statistically representative of all parcel conditions without additional sampling-design justification.

This framework is intended to support an integrated agricultural sustainability assessment for the Oglala Lakota Nation by combining geospatial analysis, ecological field evidence, and appropriate research governance.
